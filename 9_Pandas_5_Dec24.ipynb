```python
import pandas as pd
import numpy as np

data = pd.read_csv('Pfizer_1.csv')

data_melt = pd.melt(data,id_vars = ['Date', 'Drug_Name', 'Parameter'],
            var_name = "time",
            value_name = 'reading')

data_tidy = data_melt.pivot(index=['Date','time', 'Drug_Name'],
                                        columns = 'Parameter',
                                        values='reading')
data_tidy = data_tidy.reset_index()
data_tidy.columns.name = None
```

```python
def temp_mean(x):
  x['Temp_avg'] = x['Temperature'].mean()
  return x

data_tidy = data_tidy.groupby('Drug_Name', group_keys=False).apply(temp_mean)
data_tidy
```

```python
data_tidy.groupby('Drug_Name', group_keys=False).apply(temp_mean)
```

```python
data_tidy['Temperature'].fillna(data_tidy['Temp_avg'], inplace=True)
data_tidy
```

```python
def pre_mean(x):
  x['Pre_avg'] = x['Pressure'].mean()
  return x

data_tidy = data_tidy.groupby('Drug_Name', group_keys=False).apply(pre_mean)
data_tidy['Pressure'].fillna(data_tidy['Pre_avg'], inplace=True)
data_tidy
```
